<!-- Create a dictionary for common slangs and another dictionary for emojis. Implement a 
function that replaces slangs, abbreviations, and emojis with their formal counterparts. 
Normalize any irregular spaces or punctuation marks (e.g., "hello!!!" -> "hello!"). Test 
the function on a variety of text inputs containing slangs, emojis, and multiple 
punctuation.  -->


```python
import re

# Dictionary for slangs and abbreviations
slang_dict = {
    r"\bu\b": "you",
    r"\br\b": "are",
    r"\bgr8\b": "great",
    r"\blol\b": "laughing out loud",
    r"\bbrb\b": "be right back",
    r"\bidk\b": "I don't know",
    r"\bomg\b": "oh my god",
    r"\bttyl\b": "talk to you later",
    r"\bthx\b": "thanks",
    r"\bpls\b": "please"
}

# Dictionary for emojis
emoji_dict = {
    "😂": "laughing",
    "😊": "smiling",
    "❤️": "love",
    "👍": "thumbs up",
    "😢": "crying",
    "🔥": "fire",
    "🎉": "celebration",
    "💯": "perfect",
    "🙌": "praise",
    "😉": "wink"
}

def normalize_text(text):
    # Replace slangs
    for slang, formal in slang_dict.items():
        text = re.sub(slang, formal, text, flags=re.IGNORECASE)
    
    # Replace emojis
    for emoji, meaning in emoji_dict.items():
        text = text.replace(emoji, meaning)
    
    # Normalize punctuation (reduce multiple !, ?, . to one)
    text = re.sub(r'([!?.,])\1+', r'\1', text)
    
    # Normalize spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# Test cases
examples = [
    "u r gr8 😂😂!!!",
    "brb, idk what to do 😢",
    "omg!!! this party is lit 🔥🔥🔥",
    "thx 😊 ttyl!!!",
    "pls send me the doc ❤️❤️",
    "LOL!!! omg 😂😂😂!!!"
]

for e in examples:
    print("Original:", e)
    print("Normalized:", normalize_text(e))
    print()
```

Original: u r gr8 😂😂!!!
Normalized: you are great laughinglaughing!

Original: brb, idk what to do 😢
Normalized: be right back, I don't know what to do crying

Original: omg!!! this party is lit 🔥🔥🔥
Normalized: oh my god! this party is lit firefirefire

Original: thx 😊 ttyl!!!
Normalized: thanks smiling talk to you later!

Original: pls send me the doc ❤️❤️
Normalized: please send me the doc lovelove

Original: LOL!!! omg 😂😂😂!!!
Normalized: laughing out loud! oh my god laughinglaughinglaughing!

<!-- Given a sentence with mixed case, slang words, and irregular punctuation. Write a 
python program that: Replaces slang with full words, Standardizes the punctuation, 
Converts the text to lowercase.  -->


```python
import re

# Dictionary for slangs and abbreviations
slang_dict = {
    r"\bu\b": "you",
    r"\br\b": "are",
    r"\bgr8\b": "great",
    r"\blol\b": "laughing out loud",
    r"\bbrb\b": "be right back",
    r"\bidk\b": "I don't know",
    r"\bomg\b": "oh my god",
    r"\bttyl\b": "talk to you later",
    r"\bthx\b": "thanks",
    r"\bpls\b": "please"
}

def normalize_text(text):
    # Replace slangs
    for slang, formal in slang_dict.items():
        text = re.sub(slang, formal, text, flags=re.IGNORECASE)
    
    # Standardize punctuation (reduce multiple !, ?, . to one)
    text = re.sub(r'([!?.,])\1+', r'\1', text)
    
    # Normalize spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Convert to lowercase
    text = text.lower()
    
    return text

# Test cases
examples = [
    "OMG!!! U r GR8!!!",
    "brb, IDK what to do????",
    "Thx LOL!!! pls send me the doc...",
    "ttyl!!! omg omg!!!"
]

for e in examples:
    print("Original:", e)
    print("Normalized:", normalize_text(e))
    print()
```

Normalized: oh my god! you are great!

Original: brb, IDK what to do????
Normalized: be right back, i don't know what to do?

Original: Thx LOL!!! pls send me the doc...

Original: ttyl!!! omg omg!!!
